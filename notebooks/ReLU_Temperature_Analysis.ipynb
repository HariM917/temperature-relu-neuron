# Temperature Monitoring Neural Network Using ReLU

## Objective

Implement a simple artificial neuron using the ReLU (Rectified Linear Unit) activation function to process temperature data. Analyze how ReLU handles positive and negative pre-activation values and determine its effect on the representation of input values.

**Neuron Formula:** `z = w * x + b`

**ReLU Formula:** `ReLU(z) = max(0, z)`

## Step 1: Import Libraries

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully.")
````

## Step 2: Load Temperature Dataset

We load a CSV file containing temperature values in Celsius. The dataset includes both negative and positive temperatures.

````python
# Load the temperature dataset
data = pd.read_csv('../dataset/temperature_data.csv')
print(f"Loaded {len(data)} temperature values.")
````

## Step 3: Display the Dataset

````python
# Display all temperature values
print("Temperature Dataset:")
print(data)
print(f"\nTemperature range: {data['temperature_celsius'].min()}°C to {data['temperature_celsius'].max()}°C")
````

## Step 4: Define Neuron Parameters

A single neuron computes: **z = w × x + b**

Where:
- `w` (weight) = 0.5 → scales the temperature input
- `b` (bias) = -8 → shifts the pre-activation value
- `x` = temperature input

````python
# Neuron parameters (easily modifiable for viva)
weight = 0.5
bias = -8

print(f"Weight (w): {weight}")
print(f"Bias (b):   {bias}")
print(f"\nNeuron formula: z = {weight} * temperature + ({bias})")
````

## Step 5: Calculate Pre-activation

For each temperature, we calculate:
1. **Weighted input** = weight × temperature
2. **Pre-activation** = weighted input + bias

````python
# Extract temperature values
temperatures = data['temperature_celsius'].values

# Calculate weighted input and pre-activation
weighted_input = weight * temperatures
pre_activation = weighted_input + bias

print("Temperature | Weighted Input | Pre-activation (z)")
print("-" * 50)
for t, wi, z in zip(temperatures, weighted_input, pre_activation):
    print(f"{t:>11.1f}°C | {wi:>14.1f} | {z:>18.1f}")
````

## Step 6: Define ReLU Activation Function

**ReLU (Rectified Linear Unit):**

```
ReLU(z) = max(0, z)
```

- If `z < 0`: output is **0** (neuron does not fire)
- If `z >= 0`: output is **z** (neuron fires with value z)

````python
def relu(z):
    """ReLU activation function: returns max(0, z)"""
    return np.maximum(0, z)

# Test ReLU with a few examples
print("ReLU Examples:")
print(f"  ReLU(-5) = {relu(-5)}")
print(f"  ReLU(0)  = {relu(0)}")
print(f"  ReLU(7)  = {relu(7)}")
````

## Step 7: Apply ReLU to Pre-activation Values

````python
# Apply ReLU activation function
relu_output = relu(pre_activation)

print("Pre-activation (z) | ReLU Output")
print("-" * 35)
for z, r in zip(pre_activation, relu_output):
    print(f"{z:>18.1f} | {r:>11.1f}")
````

## Step 8: Complete Comparison Table

The table below shows all calculations for every temperature value.

````python
# Build the full results table
results_df = pd.DataFrame({
    'Temperature (°C)': temperatures,
    'Weighted Input': weighted_input,
    'Bias': [bias] * len(temperatures),
    'Pre-activation (z)': pre_activation,
    'ReLU Output': relu_output
})

print("\nComplete Results Table:")
print("=" * 75)
print(results_df.to_string(index=False))
print("=" * 75)
````

## Step 9: Visualize Pre-activation vs ReLU Output

This plot shows how ReLU clips negative pre-activation values to zero while passing positive values unchanged.

````python
plt.figure(figsize=(10, 6))

# Plot pre-activation
plt.plot(temperatures, pre_activation, 'bo-', label='Pre-activation (z)',
         linewidth=2, markersize=8)

# Plot ReLU output
plt.plot(temperatures, relu_output, 'rs-', label='ReLU Output',
         linewidth=2, markersize=8)

# Reference line at y=0
plt.axhline(y=0, color='gray', linestyle='--', linewidth=0.8)

plt.xlabel('Temperature (°C)', fontsize=12)
plt.ylabel('Value', fontsize=12)
plt.title('Temperature vs Pre-activation and ReLU Output', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../results/relu_activation_plot.png', dpi=150)
plt.show()
print("Plot saved to results/relu_activation_plot.png")
````

## Step 10: Bar Chart Comparison

This bar chart compares temperature, pre-activation, and ReLU output side by side for each input.

````python
x = np.arange(len(temperatures))
bar_width = 0.25

plt.figure(figsize=(12, 6))

plt.bar(x - bar_width, temperatures, bar_width, label='Temperature (°C)',
        color='steelblue')
plt.bar(x, pre_activation, bar_width, label='Pre-activation (z)',
        color='orange')
plt.bar(x + bar_width, relu_output, bar_width, label='ReLU Output',
        color='green')

plt.xlabel('Input Index', fontsize=12)
plt.ylabel('Value', fontsize=12)
plt.title('Comparison: Temperature, Pre-activation, and ReLU Output', fontsize=14)
plt.xticks(x, [f"{t:.0f}°C" for t in temperatures])
plt.legend(fontsize=11)
plt.axhline(y=0, color='gray', linestyle='--', linewidth=0.8)
plt.grid(True, axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('../results/temperature_activation_comparison.png', dpi=150)
plt.show()
print("Plot saved to results/temperature_activation_comparison.png")
````

## Step 11: Analysis of Positive and Negative Responses

Let's analyze the neuron's behavior for positive and negative pre-activation values.

````python
# Separate results by ReLU behavior
negative_mask = pre_activation < 0
positive_mask = pre_activation >= 0

print("ANALYSIS: Positive vs Negative Pre-activation")
print("=" * 55)

print("\n--- Negative Pre-activation (Neuron did NOT fire) ---")
for t, z, r in zip(temperatures[negative_mask],
                    pre_activation[negative_mask],
                    relu_output[negative_mask]):
    print(f"  Temp: {t:>6.1f}°C → z = {z:>6.1f} → ReLU = {r:.1f}")

print("\n--- Positive Pre-activation (Neuron fired) ---")
for t, z, r in zip(temperatures[positive_mask],
                    pre_activation[positive_mask],
                    relu_output[positive_mask]):
    print(f"  Temp: {t:>6.1f}°C → z = {z:>6.1f} → ReLU = {r:.1f}")

total = len(temperatures)
zero_count = negative_mask.sum()
active_count = positive_mask.sum()

print(f"\nSparsity: {zero_count}/{total} = {zero_count/total:.1%} of outputs are zero")
print(f"Active neurons: {active_count}/{total} = {active_count/total:.1%}")
````

## Step 12: Conclusion

### Key Findings:

1. **ReLU clips negative values to zero:** All temperatures that produced a negative pre-activation resulted in a ReLU output of 0. The neuron did not fire for these inputs.

2. **Positive values pass through unchanged:** Temperatures that produced positive pre-activation values had their ReLU output equal to the pre-activation. The neuron fired proportionally.

3. **Sparsity:** 5 out of 8 inputs (62.5%) produced zero output, demonstrating the sparse activation property of ReLU. In larger networks, this means only a subset of neurons are active at any time.

4. **Non-linearity:** The ReLU function creates a clear "kink" at zero — it is linear for positive values and flat at zero for negative values. This non-linearity is essential for neural networks to learn complex patterns.

5. **Threshold behavior:** With weight=0.5 and bias=-8, the neuron only activates for temperatures above approximately 16°C. The weight and bias together determine this activation threshold.

### Summary:
ReLU is a simple yet powerful activation function. Despite its simplicity (`max(0, z)`), it provides the non-linearity and sparsity that modern neural networks rely on.
